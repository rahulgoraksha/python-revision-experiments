```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# section topics:
    # 1. Scatter Plot
    # 2. Distribution Plot
    # 3. Categorical Plot
    # 4. Comparison Plot
    # 5. Seaborn Grid
    # 6. Matrix Plot
```

```python
x = [1,2,3,4,5,6,7]
y = [2,3,4,1,5,6,7]

# using pandas to convert into DataFrame
import pandas as pd

df = pd.DataFrame({'x':x, 'y':y})
df
```

### Difference between matplotlib & seabron

```python
# using matoplotlib
plt.plot(x,y)
plt.show()
```

```python
# using seaborn
sns.lineplot(x = 'x', y = 'y', data = df)
plt.show()
```

```python
# load dataset
dt = sns.load_dataset('penguins').head(20)
dt
```

### line plot using Seabron : lineplot

```python
# plot using seaborn
plt.figure(figsize=(4,3))
sns.lineplot(x = 'bill_length_mm', y = 'flipper_length_mm', data = dt, hue = 'sex', style = 'sex', palette = 'Set1', markers = ['o', '>'])

plt.title('Penguin Dataset Visualization')
plt.grid()
plt.savefig('penguin_plot.png', bbox_inches = 'tight')
plt.show()
```

### Histogram using seabron: displot

```python
df = sns.load_dataset('penguins')
df
```

```python
# plot using seaborn
plt.figure(figsize=(4,3))
sns.displot(df['bill_depth_mm'], kde = True, rug = True, color = 'red', log_scale = True) # rug : draw a kerner destimation estimation
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.displot(df['flipper_length_mm'], bins = [170,180,190, 200,210,220,230, 240], kde= True, color = 'blue') # bins eually distributed as per data range, kde : draw a smooth curve (kernel density esttimation)
plt.show()
```

```python
# Heatmap: group wise segration of data
import numpy as np
var = np.linspace(1,10,20).reshape(4,5)
var
```

```python
v = sns.heatmap(var, vmin = 0 , vmax = 12, cmap = 'YlGnBu', annot = True, annot_kws = {'size': 10, 'color': 'red'}, linewidths=10, linewidth = 2,linecolor='y', cbar = False, xticklabels=False, yticklabels=False)
v.set(xlabel = 'x-axis', ylabel = 'y-axis')
plt.show()
```

### Seaborn Styling Plot
- seaborn figure style
- removing axes spines
- scale & context

```python
df
```

#### figure style

```python
sns.boxplot(y = df['total_bill'])
plt.show()
```

```python
sns.set_style('whitegrid')
sns.barplot(y = df['total_bill'], x = df['day'], data = df)
plt.show()
```

```python
sns.set_style('whitegrid')
sns.barplot(y = df['total_bill'], x = df['day'], data = df)
plt.grid()
plt.show()
```

#### removing axes spine

```python
sns.set_style('darkgrid')
sns.barplot(y = df['total_bill'], x = df['day'], data = df)
plt.dispine() # remove the top and right spines
plt.show()
```

#### scale and context

```python
sns.set_style('darkgrid')
plt.figure(figsize=(12,3))
sns.barplot(y = df['total_bill'], x = df['day'], data = df)
sns.despine() # remove the top and right spines
plt.show()
```

```python
# context paramter 
sns.set_style('white')
sns.set_context('notebook', font_scale=1) # paper, notebook, talk, poster
plt.figure(figsize=(12,3))
sns.barplot(y = df['total_bill'], x = df['day'], data = df, palette = 'cool')
sns.despine() # remove the top and right spines
plt.show()
```
