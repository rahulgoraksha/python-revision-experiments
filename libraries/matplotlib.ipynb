```python
import matplotlib.pyplot as plt
```

### create a chart

```python
# create first chart

x = [1,2,3,4,5]
y = [10,20,25,35,45]

plt.plot(x,y) # line chart plotting
plt.show()
```

### customize chart

```python
# customize chart using dicmnentation parameters

plt.figure(figsize=(4,3)) # figure size
plt.plot(x,y, color = 'green', marker = 'o', linestyle = 'dashed', linewidth = 2, markersize = 12)
plt.title('Line Chart')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

### advanced line - mutiple line & legend

```python
# advanced line - mutiple line & legend

# data 
x = x
y1 = y
y2 = [20,30,35,45,55]

plt.figure(figsize=(4,3)) # figure size
plt.plot(x, y1, label = 'Sales 2025') # plotting y1 data
plt.plot(x, y2, label = 'Sales 2026') # plotting y2 data

plt.title('Sales Comparison')
plt.xlabel('Months')
plt.ylabel('Sales figures')

plt.legend() # shows the box representation of each line
plt.show()
```

### Bar Chart

```python
categories = ['A', 'B', 'C', 'D', 'E']
plt.bar(categories,y)
plt.title('Bar Chart')
plt.show()
```

### Histogram Chart

```python
# data for histogram used in distribution analysis
import random
data = [random.randint(1,10) for _ in range(100)]

plt.hist(data, bins = 5) # bins is the number of intervals
plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

### Pie Chart

```python
# used to show part-to-whole relationships 

# data
x = ['A','B','C','D','E']
y = [10,20,55,35,45]

plt.pie(y,labels = x, autopct = '%1.1f%%', startangle = 90) # autopct is used to show the percentage value on the pie chart
plt.title('Pie Chart Example')
plt.show()
```

### Scatter Plot

```python
# used to find the realtionships between two variables

# data
x = [10,20,25, 35,45]
y = [20,30,35,45,55]

plt.scatter(x,y, color = 'red', marker = 'x') # scatter plot
plt.title('Scatter Plot Example')
plt.show()
```

### Subplots

```python
# used to show mutiple chart in one figure

# data - Bar Chart
x = ['Mon','Tue','Wed','Thu','Fri']
y = [10,20,55,35,45]

# data - scatter plot
x_scatter = [10,20,25,35,45]
y_scatter = [20,30,35,45,55]

plt.figure(figsize=(7,4)) 
plt.subplot(1,2,1) # 1 row, 2 columns , 1st position of the chart

plt.bar(x,y, color = 'blue')
plt.title('Bar Chart')
plt.xlabel('Daily Sales')
plt.ylabel('Sales Figure')

plt.subplot(1,2,2) # 1 row, 2 columns , 2nd position of the chart

plt.scatter(x_scatter,y_scatter, color = 'green', marker = 'o')
plt.title('Scatter Plot')
plt.xlabel('User1')
plt.ylabel('User2')

plt.show()
```

### Matplotlib with pandas - real data

```python
# create df 

import pandas as pd
data = {
    'Month' : ['Jan', 'Feb', 'Mar','Apr'],
    'Sales': [12000, 11000, 13000, 25000],
}

df = pd.DataFrame(data)
df
```

```python
plt.figure(figsize = (4,3))
plt.bar(df['Month'], df['Sales'], color = 'orange')
plt.title('Monthly Sales')
plt.xlabel('Monthly Sales')
plt.ylabel('Sales Figure')
plt.show()
```

### save the chart

```python
plt.figure(figsize = (4,3))
plt.bar(df['Month'], df['Sales'], color = 'orange')
plt.title('Monthly Sales')
plt.xlabel('Monthly Sales')
plt.ylabel('Sales Figure')
# save the chart as an image file 
plt.savefig('Monthly_Sales.png', dpi = 300, bbox_inches = 'tight') # bbbox_img = 'tight'> fit entirely in the same image file

plt.show()
```

```python
# task to be performed using object oriented approach in matplotlib
import matplotlib.pyplot as plt

# create figure and axes objects
fig, ax = plt.subplots(figsize=(4,3))

# plot using axes object
ax.bar(df['Month'], df['Sales'], color='orange')

# set labels using axes methods
ax.set_title('Monthly Sales')
ax.set_xlabel('Month')
ax.set_ylabel('Sales Figure')

# save using figure object
fig.tight_layout()
fig.savefig("Monthly_Sales.png", dpi=300, bbox_inches='tight')

plt.show()
```
