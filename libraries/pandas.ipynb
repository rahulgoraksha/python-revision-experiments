```python
df = pd.DataFrame([11,22,33], columns= ["Col_Name"])
print(df)
#          column
#   row      0
#    0.      11
#    1.      22
#    2.      33

print(type(df))
```

### creating Dataframe

```python
data = {
    'Name' : ["Rahul", "Janvi","Aishwarya","Simran"],
    "Age" : [16,17,18,19],
    "Salary" : [90000, 70000, 50000, 30000]
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```
