### creating Dataframe

```python
data = {
    'Name' : ["Rahul", "Janvi","Aishwarya","Simran"],
    "Age" : [16,17,18,19],
    "Salary" : [90000, 70000, 50000, 30000]
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

### Rows and Column Selection

```python
# select column

df['Name'] # single column
```

```python
df[['Name','Salary']]   # mutiple columns
```

```python
# slect rows. : loc = label based index 
df.loc[df.Name=="Janvi"]    # location index based on name

df.loc[(df.Name=='Rahul') & (df.Salary>= 50000)]
```

```python
df.loc[0:2]
```

```python
# iloc - index-value based index
df.iloc[2]
df.iloc[0:2] # start -stop-step
```

### Row and Columns - Operation (Add, update, delete)

#### Add

```python
df
```

```python
# add new column

df['Team'] = ["CEO", "MD", "MIS", "DA"]

df
```

```python
# add new column

df['Bonus'] = df['Salary']*0.2
print(df)


# add new row
df.loc[len(df)] = ['ABC',21,21000,'IT',2000]
print(df)
```

#### update

```python
# update value

df.loc[0,'Salary'] = 74000
df
```

```python
# update value - using column value
df.loc[df.Name == "Rahul","Salary"] = 1000000
df
```

### Delete

```python
# delete value - row and column values
# axis = 0 (row), 1 (column)

df.drop(df[df['Name']== 'ABC'].index) # index used to capture entire row 
df
```

```python
df[df.Name == 'ABC']
```

```python
df.drop('Bonus', axis = 1) # dop column
```

```python
# drop multiple columns
df.drop(['Bonus','Team'], axis = 1)
```
