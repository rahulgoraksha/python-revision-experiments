### Concatenate and merge DataFrame(Joins)

```python
df1 = pd.DataFrame({
    'ID':[1,2,3],
    'Name' : ['A','B','C']
})
df1
```

```python
df2 = pd.DataFrame({
    'ID':[1,2,2,4],
    'Score' : [88,96,77,79]
})
df2
```

```python
# concatenate the created DataFrame

pd.concat([df1,df2],axis = 0)   # vertical table join/ row level / top on top
```

```python
pd.concat([df1,df2],axis = 1) # horizontal table join/ column level / side on side
```

```python
# merge table

pd.merge(df1, df2, how = "inner", on="ID")  
```

```python
# create a realistic sales dataset you can practice many pivot questions on.

import pandas as pd

data = {
    "Date": [
        "2024-01-01","2024-01-02","2024-01-03","2024-01-04","2024-01-05",
        "2024-01-06","2024-01-07","2024-01-08","2024-01-09","2024-01-10",
        "2024-01-11","2024-01-12","2024-01-13","2024-01-14","2024-01-15"
    ],
    "City": [
        "Mumbai","Mumbai","Delhi","Delhi","Pune",
        "Pune","Mumbai","Delhi","Pune","Mumbai",
        "Delhi","Mumbai","Pune","Delhi","Mumbai"
    ],
    "Product": [
        "Laptop","Mobile","Laptop","Tablet","Mobile",
        "Laptop","Tablet","Mobile","Tablet","Laptop",
        "Laptop","Mobile","Laptop","Tablet","Mobile"
    ],
    "Salesperson": [
        "Rahul","Amit","Rahul","Sneha","Amit",
        "Sneha","Rahul","Amit","Sneha","Rahul",
        "Amit","Sneha","Rahul","Amit","Sneha"
    ],
    "Units": [5,8,4,6,7,3,5,9,4,6,8,7,2,5,9],
    "Revenue": [50000,80000,40000,36000,56000,30000,25000,90000,28000,60000,80000,70000,20000,30000,90000]
}

df = pd.DataFrame(data)
df["Date"] = pd.to_datetime(df["Date"])

df
```

```python
pd.pivot_table(df, values="Revenue", index="City", aggfunc="sum")
```

```python
pd.pivot_table(df, values="Revenue", index="City", columns="Product", aggfunc="sum")
```

```python
pd.pivot_table(df, values="Units", index="Salesperson", aggfunc="count")
```

```python
pd.pivot_table(df, values="Units", index="Product", aggfunc="mean")
```

```python
pd.pivot_table(
    df,
    values="Revenue",
    index="City",
    columns="Product",
    aggfunc=["sum","mean","count"]
)
```

```python
pd.pivot_table(
    df,
    values="Revenue",
    index=df["Date"].dt.month,
    columns="City",
    aggfunc="sum"
)
```
