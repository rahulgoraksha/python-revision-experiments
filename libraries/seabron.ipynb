```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# section topics:
    # 1. Scatter Plot
    # 2. Distribution Plot
    # 3. Categorical Plot
    # 4. Comparison Plot
    # 5. Seaborn Grid
    # 6. Matrix Plot
```

```python
x = [1,2,3,4,5,6,7]
y = [2,3,4,1,5,6,7]

# using pandas to convert into DataFrame
import pandas as pd

df = pd.DataFrame({'x':x, 'y':y})
df
```

### Difference between matplotlib & seabron

```python
# using matoplotlib
plt.plot(x,y)
plt.show()
```

```python
# using seaborn
sns.lineplot(x = 'x', y = 'y', data = df)
plt.show()
```

```python
# load dataset
dt = sns.load_dataset('penguins').head(20)
dt
```

### line plot using Seabron : lineplot

```python
# plot using seaborn
plt.figure(figsize=(4,3))
sns.lineplot(x = 'bill_length_mm', y = 'flipper_length_mm', data = dt, hue = 'sex', style = 'sex', palette = 'Set1', markers = ['o', '>'])

plt.title('Penguin Dataset Visualization')
plt.grid()
plt.savefig('penguin_plot.png', bbox_inches = 'tight')
plt.show()
```

### Histogram using seabron: displot

```python
df = sns.load_dataset('penguins')
df
```

```python
# plot using seaborn
plt.figure(figsize=(4,3))
sns.displot(df['bill_depth_mm'], kde = True, rug = True, color = 'red', log_scale = True) # rug : draw a kerner destimation estimation
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.displot(df['flipper_length_mm'], bins = [170,180,190, 200,210,220,230, 240], kde= True, color = 'blue') # bins eually distributed as per data range, kde : draw a smooth curve (kernel density esttimation)
plt.show()
```

### Barplot using seaborn

```python
df
```

```python
sns.barplot(x = 'island', y = 'bill_length_mm', data = df, hue = 'sex')
plt.show()
```

```python
# wish to rearrange the order of the bar
order_1 = ['Dream', 'Biscoe', 'Torgersen']
sns.barplot(x = 'island', y = 'bill_length_mm', data = df, hue = 'sex', order = order_1, hue_order=['Female', 'Male'], orient = 'v', saturation=10, errcolor='y', errwidth = 5)
plt.show()
```

```python
# if you want orientation of bar in horizontal then the all data should be in the number format

# wish to rearrange the order of the bar
order_1 = ['Dream', 'Biscoe', 'Torgersen']
sns.barplot(x = 'island', y = 'bill_length_mm', data = df, orient = 'h')
plt.show()
```
