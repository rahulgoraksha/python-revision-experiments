# Key Understanding About Python

**Python Code Execution Steps**

1. **Lexical Analysis**  
   The Python interpreter breaks the source code into smaller parts called **tokens**, such as keywords, identifiers, numbers, symbols, and punctuation.

2. **Syntax Parsing**  
   Python checks whether the structure of the code is correct. If the syntax rules are violated, a **SyntaxError** is raised.

3. **Bytecode Generation**  
   After successful syntax checking, the code is converted into a simpler set of instructions called **bytecode**.

4. **Execution by PVM**  
   The **Python Virtual Machine (PVM)** executes the bytecode and performs operations like calculations, value assignments, and function calls.

5. **Error Handling and Output**  
   During execution, Python handles runtime errors if any occur and displays the output such as printed messages or returned values.


**Interpretor**
1. It tends to translate and execute code line by line.
2. Execution: Line by line
3. Speed: Slower
4. Debugging: Easier
5. Example: Python, JavaScript, Ruby and PHP

**Compiler**
1. It tends to translate the entire code into machine code before program runs
2. Execution: Entire program at once
3. Speed: Faster
4. Debugging: Complex
5. Example: C, C++, Java

## Loops 
1. It enable you to perform repetitive tasks efficiently without writing redundant code.
2. They iterate over a sequence (like a list, tuple, string, or range) or execute a block of code as long as a specific **condition is met**
3. Range is kind of similiar to slice often used with the for loop

- Types of Loops

  1. **While Loop**
     - It repeatedly executes a block of code as long as given condition remains True

  2. **For Loop**
        - It iterate over a sequence (such as list, tuple, dictionary, set, or string) and excute a block of code for each element in the sequence.

  3. **Nested Loop**

- **Loop Control Statements**
1. It allows you to **alter** the normal flow of a loop
    - pass statement: used as a placeholder(it does nothing) for the future code {already covered in function}
    - break statement: act as terminates the loop entireley, exiting from it immediately
    - continue statement: skips the current iterations and moves to the next one

```python
# While Loop Example
# print number from 0 to 3

count = 0           # initializing count variable
while count <= 3:    #condition to check if count is less than or equal to 3
    print(count)
    count += 1
```

```python
# print number form 1 to 5 using while loop

count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
# print number from 5 to 1 using while loop

count = 5
while count >= 1:
    print(count)
    count -= 1
else:
    print("While loop has ended.")
```

## Nested Loop

- Loop inside another loop is nested loop. It means every single time the outer loop runs, the inner loop runs all of its iterations.

**Why Use Nested loops?**
1. Handling the multi dimensional data: such as matrices, grids, list
2. Complex Iterations: operation depend on mutiple variables dimension
3. Pattern Generation: creating pattern, such as graphics or games.

```python
# Nested Loops Example: understanding 

for i in range(1, 4):
    print(i)
    i += 1

for i in range(1, 4):
    print(i)
    i += 1

for i in range(1, 4):
    print(i)
    i += 1
```

```python
# Nested Loops Example

for i in range(3):
    print("Outer loop iteration:", i)
    for j in range(1,4):
        print(j)
    print("---")
```

```python
# Nested loop Example using While Loop & For Loop

i = 1
while i < 4:
    print("Outer loop iteration:", i)
    for j in range(1,4):
        print(j)
    print("---")
    i += 1
```

```python
# print prime numbers between range of 2  to 10

for num in range(2,10):
    for i in range(2, num):
        if num % i == 0:
            break
    else:
            print(num)
```

```python
# List Comprehensions: provide a concise way to create a list, consist of brackets containing an expression followed by a clause, and optionally if clause

# list of square
squares = [x**2 for x in range(1,6)]
print(squares)

# list of even number 
even_number = [x for x in range(1,11) if x % 2 ==0]
print(even_number)

# applying function to each element : iterating over a list
fruits = ["apple", "banana", "cherry"]
uppercase_fruits = [fruit.upper() for fruit in fruits]
print(uppercase_fruits)

# flatten a nested list using a list comprehension : {flatten : convert a nested list to normalised list}
nest_list = [[1,2],[3,4],[5,6]]

def flatten(lst):
    return [item for sublist in lst for item in sublist]

final_list = flatten(nest_list)
print(final_list)

```
