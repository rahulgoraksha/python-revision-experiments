```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Term_Insurance_Data = {
    'Age' : [25,30,35,40,45],
    'Premium': [18000,32000, 42000, 47000, 55000]
}

print(Term_Insurance_Data)
```

```python
df = pd.DataFrame(Term_Insurance_Data)
df
```

- we'll find out the premium whose age is 21 and 50 ?

```python
df
```

```python
x = df['Age']
y = df['Premium']
```

```python
plt.figure(figsize=(4,3))
plt.scatter(x,y)
plt.xlabel('Age')
plt.ylabel('Premium')
plt.show()
```

### lets begin for plotting..... linear Regression

```python
import pandas as pd
import seaborn as sns
from sklearn import linear_model
```

```python
df = pd.DataFrame(Term_Insurance_Data)
df
```

```python
sns.lmplot(x = 'Age',y = 'Premium',data= df)
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['Age']],df[['Premium']])
```

```python
reg.predict([[21]])
```

```python
reg.predict([[50]])
```

```python
reg.coef_
```

```python
reg.intercept_
```
