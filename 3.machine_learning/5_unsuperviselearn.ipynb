### UnSupervised Machine Learning

- In this mainly, you train the machine using the data which is un-labeled and models itself find the hidden patterns and insight from the given data.





### Type	Purpose
- Clustering: 	Group similar data
- Dimensionality: Reduction	Reduce features
- Association:	Find item relationships
- Anomaly Detection:	Find outliers

### Unsupervised leanrning Algorithm
1. K-Means Clustering
2. KNN (K-Nearest Neighbour)
3. Hierarchical Clustering
4. Neural Networks/ Deep Learning
5. Single Value Decomposition
6. Distribution Models
7. Principal Component Analysis
8. Apriori Algortithm

### Advantage:
- used for more complex tasks
- helpful in finding pattern in data
- saves lot of mannual work and expenses

### Disadvantage:
- less accuracy
- time-consuming
- more the features, more the complexity
