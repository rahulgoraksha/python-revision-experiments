### Machine learning Life Cycle
1. Data Collection: Gather raw data from databases, APIs, sensors, files, etc.
2. Data Preparation: Clean basic issues like missing values, duplicates, formatting.
3. Data Wrangling: Transform and engineer features to make data ML-ready.
4. Data Modelling: Choose the suitable ML algorithm for the problem.
5. Model Training: Train the model using training data.
6. Model testing: Evaluate model performance using test/validation data.
7. Deployment: Deploy the model to real-world applications.

### AI v/s ML v/s DL

- (((DL) > ML) > AL)
- AI: making computer capable enough to perform the task
- ML: enables machine to learn from past dataset without being explicitly programmed
    - supervised : label dataset
    - unsupervised : unlabel dataset
    - reinforcement : बक्षीस आणि शिक्षा यावरून शिकणे
- DL: algorithm inspired by the structure and function of human brains (neural network)
