### crossvalidation:

- evaluating model performance

- it mostly tells which moel goes best

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_digits
```

```python
df = load_digits()
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.data, df.target, test_size=0.3)
```

```python
model1 = LogisticRegression()
model2 = SVC()
model3 = RandomForestClassifier(n_estimators=100)
```

```python
model1.fit(X_train, y_train)
model1.score(X_test, y_test)
```

```python
model2.fit(X_train, y_train)
model2.score(X_test, y_test)
```

```python
model3.fit(X_train, y_train)
model3.score(X_test, y_test)
```

# while running all these seperately so now crossvalidation comes into picture

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(LogisticRegression(), df.data, df.target,  cv=3)
```

```python
cross_val_score(SVC(), df.data, df.target, cv=3)
```

```python
cross_val_score(RandomForestClassifier(), df.data, df.target, cv=3)
```
