### PCA 

- oftenly used to reduce the dimensionality of the data
- PCA is called dimensionality reduction technique as it can help us reduce dimensions

- PCA: principal component analysis
    - is a process of figuring out most important features or principal components that has the most impact on the target variable

- few things to keep in mind before using PCA
1. scale feature before applying PCA
2. Accuracy might drop

```python
X = df.data
y = df.target
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled,y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

model1 = LogisticRegression()
model1.fit(X_train,y_train)
model1.score(X_test,y_test)
```

### using PCA

```python
from sklearn.decomposition import PCA

pca = PCA(0.95)

x_pca = pca.fit_transform(X)
x_pca.shape
```

```python
x_pca
```

```python
pca.explained_variance_ratio_ #tells newly generated 29 component as said in shape
```

```python
pca.n_components
```

```python
X_train_pca, X_test_pca, y_train, y_test = train_test_split(x_pca,y, test_size=0.2, random_state=  42)
```

```python
from sklearn.linear_model import LogisticRegression

model2 = LogisticRegression(max_iter=1000)
model2.fit(X_train_pca,y_train)
model2.score(X_test_pca,y_test)
```
