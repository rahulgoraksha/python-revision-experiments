```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
insurance_data = {
    "age": [22, 25, 47, 52, 46, 56, 55, 60, 62, 61,
            18, 28, 27, 29, 49, 55, 25, 58, 19],
    
    "bought_insurance": ["No", "No", "Yes", "Yes", "Yes",
                         "Yes", "Yes", "Yes", "Yes", "Yes",
                         "No", "No", "No", "No", "Yes",
                         "Yes", "No", "Yes", "No"]
}

print(insurance_data)
```

```python
df = pd.DataFrame(insurance_data)
df
```

```python
df['bought_insurance'] = df['bought_insurance'].map({'No': 0, 'Yes': 1})
df
```

```python
X = df[['age']]
```

```python
y = df[['bought_insurance']]
```

```python
plt.scatter(x = 'age', y = 'bought_insurance', data = df)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.predict(X_test)
```

```python
lr.predict([[60]])
```
