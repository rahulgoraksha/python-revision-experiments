### Supervised Machine Learning

- In this mainly, you train the machine using the data which is labeled i.e some inout data is already tagged with correct answer and this algorithm learns from labeled training data that helps you to predict the further outcomes



# Types of supervise ML Algorithm

1. Regression > {output: numeric}

2. Classification > {output: category}
- Random Forest
- Decision Trees
- Logistics Regression
- Support Vector Machines

- you can easily test and debug the model

### Disadvantage:
